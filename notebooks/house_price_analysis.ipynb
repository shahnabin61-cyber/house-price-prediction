- Prepare the dataset for modeling and prediction

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_house_price = pd.read_csv('../data/train.csv')

# FIRST LOOK 

print("Shape:", df_house_price.shape)
print("\nColumn:", df_house_price.columns.tolist())
print("\n -----FIRST 10 ROWS----")
df_house_price.head(10)
```

### 1. Data Overview

This section provides an initial inspection of the training dataset, including table shape, sample rows, and column summaries.


### 2. Data Quality and Missing Values

This section evaluates data completeness, checks for missing values, and summarizes the key numerical features.


```python
import numpy as np

missing_summary = df_house_price.isna().sum().sort_values(ascending=False)
missing_summary = missing_summary[missing_summary > 0]

print("Missing values by column:")
print(missing_summary)

print("\nNumerical summary:")
df_house_price.describe().T
```

### 3. Summary Statistics and Feature Distributions

This section highlights key numerical feature distributions and identifies the strongest predictors for `SalePrice`.


```python
numeric_features = df_house_price.select_dtypes(include=["number"]).copy()

feature_corr = numeric_features.corr()["SalePrice"].abs().sort_values(ascending=False)
print("Top 10 features most correlated with SalePrice:")
print(feature_corr.head(10))

plt.figure(figsize=(10, 6))
sns.histplot(df_house_price["SalePrice"], kde=True, color="steelblue")
plt.title("SalePrice Distribution")
plt.xlabel("SalePrice")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(12, 10))
selected = feature_corr.head(9).index.tolist()
sns.heatmap(df_house_price[selected].corr(), annot=True, fmt=".2f", cmap="coolwarm", cbar=False)
plt.title("Top Numeric Feature Correlations")
plt.show()

### 4. Recommended Next Steps

- Impute missing values for high-impact features
- Encode categorical variables with strong `SalePrice` correlation
- Engineer new features like total living area and building age
- Split data and build a validation-ready model pipeline
```
